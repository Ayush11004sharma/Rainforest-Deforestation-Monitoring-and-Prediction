# 01 — Data Inspection & Validation

**Objective:** Load and inspect all four raw CSV datasets from the Global Forest Change (GFW) project.
Verify schemas, data types, missing values, geographic coverage, and year ranges.

**Source:** Hansen et al. (2013), University of Maryland / GLAD Lab, derived from Landsat satellite imagery.


```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
from config import *
from src.data_loader import load_all_datasets, validate_dataset
```

## 1. Load All Datasets


```python
datasets = load_all_datasets()
for key, df in datasets.items():
    if df is not None:
        print(f'=== {key} ===')
        print(f'Shape: {df.shape}')
        print(f'Columns: {list(df.columns)}')
        print(f'Missing values: {df.isnull().sum().sum()}')
        print(f'Duplicates: {df.duplicated().sum()}')
        print()
```

## 2. Inspect Primary Dataset: Subnational Tree Cover Loss


```python
df = datasets['subnational_loss']
print('Shape:', df.shape)
print('Dtypes:')
print(df.dtypes)
print()
print('Unique thresholds:', sorted(df['threshold'].unique()))
print('Unique countries:', df['country'].nunique())
df.head(3)
```

## 3. Brazil — Subnational Regions at Threshold=30


```python
br = df[(df['country'] == 'Brazil') & (df['threshold'] == 30)]
print(f'Brazil rows at threshold=30: {len(br)}')
print(f'Regions ({br["subnational1"].nunique()}):')
for r in sorted(br['subnational1'].unique()):
    print(f'  - {r}')
print()
br[['subnational1', 'extent_2000_ha', 'extent_2010_ha', 'tc_loss_ha_2001', 'tc_loss_ha_2022']]
```

## 4. Inspect Carbon Data


```python
dc = datasets['subnational_carbon']
print('Shape:', dc.shape)
print('Columns:', list(dc.columns)[:10], '...')
print()
brc = dc[(dc['country'] == 'Brazil') & (dc['umd_tree_cover_density_2000__threshold'] == 30)]
print(f'Brazil carbon rows (threshold=30): {len(brc)}')
print(f'Non-null emissions 2001: {brc["gfw_forest_carbon_gross_emissions_2001__Mg_CO2e"].notna().sum()}')
brc[['subnational1', 'gfw_forest_carbon_gross_emissions_2001__Mg_CO2e',
     'gfw_forest_carbon_gross_emissions_2022__Mg_CO2e']].head(10)
```

## 5. Data Quality Report


```python
for key, df in datasets.items():
    report = validate_dataset(df, key)
    print(f'--- {key} ---')
    print(f'  Shape: {report["shape"]}')
    print(f'  Missing: {report["total_missing"]}')
    print(f'  Duplicates: {report["duplicate_rows"]}')
    print(f'  Year range: {report.get("year_range", "N/A")}')
    print()
```

## 6. Country-Level Data


```python
cl = datasets['country_loss']
print('Country loss shape:', cl.shape)
br_cl = cl[(cl['country'] == 'Brazil') & (cl['threshold'] == 30)]
print('Brazil (threshold=30):')
print(br_cl[['country', 'extent_2000_ha', 'tc_loss_ha_2001', 'tc_loss_ha_2022']])
```

## Conclusions

- **4 datasets** loaded successfully with no file errors
- **Subnational loss:** 28,000 rows × 29 columns — no missing values
- **Carbon data:** Significant missingness outside countries with large forests
- **Brazil:** 27 subnational regions available at threshold=30
- **Years:** 2001–2022 (22 annual observations per region)
- **Units:** Tree-cover loss in hectares; carbon in Mg CO₂e
- **Threshold=30** means pixels with ≥30% tree-cover density in year 2000
